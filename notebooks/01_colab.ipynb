## 0. Metodología de creación de la base de datos

La consolidación de la base de datos (`base_final.xlsx`) se estructuró en cuatro fases: **definición de variables**, **recopilación de información**, **estandarización de datos** y **estructuración relacional**.

### 0.1. Definición de parámetros y alcance

La selección de variables se basó en los estándares del proyecto **SAG (Sistema Acuífero Guaraní)**, validados técnicamente en los cuatro países que comparten el acuífero. Referencia principal: *"Proyecto Porte Medio para la ejecución del Plan de Acciones Estratégicas (PAE) del Sistema Acuífero Guaraní (SAG) — Implementación del programa de acción estratégica del Acuífero Guaraní: posibilitar acciones regionales a través del Comité de Monitoreo y Modelado (CMM)"*.

### 0.2. Recopilación de fuentes e información

Se relevaron instituciones públicas y privadas para identificar pozos en las dos áreas piloto:

- **Organismos públicos:** OSE, DINAGUA, DINAMIGE (Izquierdo y Techera 2019-2023, 2022, 2024), ANII (ANII 2022), MGAP (PRENADER 1994).
- **Academia:** tesis de grado (Montaño 2005; Mañay et al. 2014) y posgrado (Collazo 2006; Carrión 2023), publicaciones (Veroslavsky et al. 2024; Carballo et al. 1995; Spoturno y Oyhantçabal 2004) y proyectos (Postiglione et al. 2009; Proyecto RLA/8/031 2005; Manganelli et al. 2022).

De esta integración se identificaron aproximadamente **1774 pozos** con información relevante.

### 0.3. Normalización y curación de datos

- **Unificación de nomenclatura:** se creó el **"Código UK"**, identificador único por pozo que actúa como llave foránea (*foreign key*) entre todas las tablas.
- **Corrección de errores:** depuración de errores de tipeo, validación de coordenadas geográficas, unificación de unidades de medida.
- **Descripción geológica:** se añadió información de formaciones geológicas mediante reinterpretación de las descripciones litológicas originales (Bossi et al. 1966, 1975; Loureiro 2019; Spoturno et al. 2004; Veroslavsky et al. 2022, 2024).

### 0.4. Estructura de la base de datos

### 0.5. Área de estudio

El **Sistema Acuífero Raigón (SAR)** se ubica al sur del departamento de San José. El sector continuo se extiende entre el arroyo Pavón y los afloramientos de rocas cristalinas (oeste), el río San José y las rocas cristalinas (norte), el arroyo de la Virgen y el río Santa Lucía (este), y el Río de la Plata (sur). Totaliza un área cercana a **2300 km²**.

*Fuente: De los Santos Gregoraschuk et al. (2019). Gestión Ambiental del Sistema Acuífero Raigón (Serie FPTA N.º 78). INIA, Montevideo. ISBN 978-9974-38-427-9.*

> **Nota metodológica — cobertura geográfica de la base PUK:** la base principal de pozos (`base_final.xlsx`) contiene registros de dos áreas piloto distintas (Raigón/San José y Tacuarembó). Se comprobó que la columna `Departamento` **no es confiable** como criterio para delimitar cada acuífero (mezcla de mayúsculas/minúsculas, departamentos limítrofes, ~27 pozos con el departamento equivocado según sus propias coordenadas). El filtro espacial correcto es el **recorte por el límite geométrico** de cada área (`LimiteCP` para el SAR, `Area_Piloto_Tacuarembo_entera` para Guaraní) — así se implementó en este proyecto (`cleaning.clasificar_acuifero_por_ubicacion`), con la latitud y el Departamento como respaldo solo para pozos fuera de esos polígonos o sin coordenadas válidas.

## 3. Configurar ngrok
Necesitás una cuenta gratuita en https://dashboard.ngrok.com/ y tu authtoken (el mismo tipo de token que usaste para Libertad — **no lo compartas**, funciona como una contraseña).

Pegá tu token reemplazando el texto de abajo antes de correr la celda.

```
from pyngrok import ngrok

ngrok.kill()  # por si quedó una sesión previa abierta
ngrok.set_auth_token("PEGA_ACA_TU_TOKEN")
```

## 4. Levantar Streamlit + túnel

```
import subprocess, time

proceso = subprocess.Popen([
    "streamlit", "run", "/content/raigon_guarani/app/streamlit_app.py",
    "--server.port", "8501", "--server.headless", "true",
])
time.sleep(6)

tunel = ngrok.connect(8501)
print("Abrí esta URL en el navegador:", tunel)
```

## 5. Cuando termines
Corré esta celda para cerrar el túnel y liberar el puerto (si vas a volver a correr la Celda 4, hacelo antes).

```
ngrok.kill()
proceso.terminate()
print("Túnel y app detenidos.")
```
